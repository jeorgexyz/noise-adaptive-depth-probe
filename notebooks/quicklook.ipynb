# Quick-Look: Exploratory Data Analysis

Interactive notebook for inspecting `results/depth_threshold_results.json`.

No IBM Quantum credentials required — works entirely from saved data.

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.io_utils import load_results

RESULTS_FILE = project_root / 'results' / 'depth_threshold_results.json'
results = load_results(str(RESULTS_FILE))
print('Loaded:', RESULTS_FILE)
print('Timestamp:', results['metadata']['timestamp'])
print('Backends:', results['metadata']['backends'])
print('Qubits:',   results['metadata']['n_qubits'])
print('Depths:',   results['metadata']['depths'])
print('Shots:',    results['metadata']['shots'])
```

## 1. Build a Tidy DataFrame

```python
rows = []
for backend, bdata in results['data'].items():
    for depth_key, ddata in bdata.items():
        rows.append({
            'backend':          backend,
            'depth_logical':    ddata['circuit_depth_logical'],
            'depth_transpiled': ddata['circuit_depth_transpiled'],
            'gate_count':       ddata['gate_count'],
            'cz_gates':         ddata['cz_count'],
            'two_qubit_gates':  ddata['two_qubit_gates'],
            'ghz_fidelity':     ddata['metrics']['ghz_fidelity'],
            'parity':           ddata['metrics']['parity_expectation'],
            'hog':              ddata['metrics']['heavy_output_frequency'],
            'job_id':           ddata['job_id'],
        })

df = pd.DataFrame(rows).sort_values('depth_logical').reset_index(drop=True)
print(df[['depth_logical','depth_transpiled','cz_gates',
          'ghz_fidelity','parity','hog']].to_string(index=False))
```

## 2. All Three Metrics vs Depth

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for backend, grp in df.groupby('backend'):
    axes[0].plot(grp['depth_logical'], grp['ghz_fidelity'],  'o-', lw=2, ms=7, label=backend)
    axes[1].plot(grp['depth_logical'], grp['parity'],         's-', lw=2, ms=7, label=backend)
    axes[2].plot(grp['depth_logical'], grp['hog'],            '^-', lw=2, ms=7, label=backend)

axes[0].set(title='GHZ Fidelity',          xlabel='Depth', ylabel='Fidelity',                ylim=[0.75, 1.0])
axes[1].set(title='Parity Expectation',    xlabel='Depth', ylabel='⟨Z⊗Z⊗...⊗Z⟩')
axes[1].axhline(0, color='k', linestyle='--', alpha=0.3)
axes[2].set(title='Heavy Output Frequency',xlabel='Depth', ylabel='HOG',                     ylim=[0.75, 1.0])

for ax in axes:
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.set_xticks(sorted(df['depth_logical'].unique()))

n = results['metadata']['n_qubits']
plt.suptitle(f'Results: {n}q on ibm_torino', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 3. Raw Bitstring Distributions (first and last depth)

```python
backend = list(results['data'].keys())[0]
depth_keys = sorted(results['data'][backend].keys(), key=lambda k: int(k.split('_')[1]))
n_q = results['metadata']['n_qubits']
targets = {'0' * n_q, '1' * n_q}

for dk in [depth_keys[0], depth_keys[-1]]:
    counts = results['data'][backend][dk]['counts']
    depth  = results['data'][backend][dk]['circuit_depth_logical']
    total  = sum(counts.values())

    sorted_counts = sorted(counts.items(), key=lambda x: -x[1])[:12]
    labels, vals  = zip(*sorted_counts)
    probs = [v / total for v in vals]
    colors = ['green' if l in targets else 'steelblue' for l in labels]

    fig, ax = plt.subplots(figsize=(10, 3))
    ax.bar(labels, probs, color=colors)
    ax.set_title(f'Depth {depth} — top-12 bitstrings (green = target GHZ states)', fontsize=12)
    ax.set_ylabel('Probability')
    ax.set_xlabel('Bitstring')
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## 4. Fidelity vs Transpiled Depth and Gate Count

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

sc1 = ax1.scatter(df['depth_transpiled'], df['ghz_fidelity'],
                  s=80, c=df['depth_logical'], cmap='viridis')
ax1.set(xlabel='Transpiled Depth', ylabel='GHZ Fidelity', title='Fidelity vs Transpiled Depth')
ax1.grid(True, alpha=0.3)
plt.colorbar(sc1, ax=ax1, label='Logical Depth')

sc2 = ax2.scatter(df['cz_gates'], df['ghz_fidelity'],
                  s=80, c=df['depth_logical'], cmap='viridis')
ax2.set(xlabel='CZ Gate Count', ylabel='GHZ Fidelity', title='Fidelity vs CZ Gate Count')
ax2.grid(True, alpha=0.3)
plt.colorbar(sc2, ax=ax2, label='Logical Depth')

plt.tight_layout()
plt.show()
```

## 5. Summary Statistics

```python
print('=== Summary Statistics ===')
print(df[['ghz_fidelity','parity','hog']].describe().round(4))

print('\n=== Fidelity at Each Depth ===')
for _, row in df.iterrows():
    tag = ' <-- recommended max' if row['depth_logical'] == 10 else ''
    print(f"  depth {int(row['depth_logical']):>2}: fidelity={row['ghz_fidelity']:.3f}{tag}")

total_decay = df['ghz_fidelity'].iloc[0] - df['ghz_fidelity'].iloc[-1]
print(f'\nTotal fidelity decay (depth 2 → 20): {total_decay:.3f} ({total_decay*100:.1f}%)')

rate = np.polyfit(df['depth_logical'], df['ghz_fidelity'], 1)[0]
print(f'Linear decay rate: {rate*100:.3f}% per depth step')
```
