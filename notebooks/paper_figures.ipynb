# Publication Figures

Generates all figures used in the paper from the saved experimental results.

**Output files** (written to `results/` and `paper/figures/`):
- `publication_figure.png` — main 4q vs 5q fidelity comparison (Figure 1)
- `gate_count_analysis.png` — fidelity vs gate count + scaling (Figure 2)
- `depth_threshold_analysis.png` — 3-panel fidelity/parity/HOG (Supplementary)

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import json
import numpy as np
import matplotlib.pyplot as plt
import shutil

from src.io_utils import load_results
from src.plotting import plot_results

RESULTS_FILE = project_root / 'results' / 'depth_threshold_results.json'
FIGURES_DIR  = project_root / 'paper' / 'figures'
RESULTS_DIR  = project_root / 'results'
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

DATA_4Q = {
    'depths':     [2, 4, 6, 8, 10, 15, 20],
    'fidelities': [0.916, 0.928, 0.908, 0.915, 0.879, 0.882, 0.884],
    'cz_gates':   [6, 12, 18, 24, 30, 45, 60],
}
DATA_5Q = {
    'depths':     [2, 4, 6, 8, 10, 15, 20],
    'fidelities': [0.894, 0.905, 0.879, 0.874, 0.863, 0.851, 0.837],
    'cz_gates':   [8, 16, 24, 32, 40, 60, 80],
}

print('Project root:', project_root)
print('Results file exists:', RESULTS_FILE.exists())
```

## Figure 1 — Main Publication Figure: 4q vs 5q Fidelity Comparison

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(DATA_4Q['depths'], DATA_4Q['fidelities'],
        'o-', color='steelblue', linewidth=2.5, markersize=8,
        label='4-qubit (plateau ≈ 88%)')
ax.plot(DATA_5Q['depths'], DATA_5Q['fidelities'],
        's--', color='darkorange', linewidth=2.5, markersize=8,
        label='5-qubit (linear decay)')

ax.axhline(0.85, color='red',   linestyle=':', linewidth=1.5, alpha=0.7, label='85% threshold')
ax.axhline(0.90, color='green', linestyle=':', linewidth=1.5, alpha=0.7, label='90% threshold')
ax.axvline(10,   color='gray',  linestyle='--', linewidth=1.5, alpha=0.5, label='Recommended max depth')

ax.set_xlabel('Circuit Depth (logical layers)', fontsize=13)
ax.set_ylabel('GHZ State Fidelity', fontsize=13)
ax.set_title('Fidelity Decay vs Circuit Depth\nibm_torino (Heron r2), Jan 2026',
             fontsize=13, fontweight='bold')
ax.set_ylim(0.80, 0.96)
ax.set_xticks(DATA_4Q['depths'])
ax.legend(fontsize=10, loc='lower left')
ax.grid(True, alpha=0.3)

plt.tight_layout()
out = FIGURES_DIR / 'publication_figure.png'
plt.savefig(out, dpi=300, bbox_inches='tight')
plt.savefig(RESULTS_DIR / 'publication_figure.png', dpi=300, bbox_inches='tight')
print('Saved:', out)
plt.show()
```

## Figure 2 — Gate Count Analysis

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(DATA_4Q['cz_gates'], DATA_4Q['fidelities'],
         'o-', color='steelblue', linewidth=2, markersize=8, label='4-qubit')
ax1.plot(DATA_5Q['cz_gates'], DATA_5Q['fidelities'],
         's--', color='darkorange', linewidth=2, markersize=8, label='5-qubit')
ax1.set(xlabel='Number of CZ Gates', ylabel='GHZ Fidelity',
        title='Fidelity vs Gate Count', ylim=[0.80, 0.96])
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

depths_arr = np.array(DATA_4Q['depths'])
ax2.plot(DATA_4Q['depths'], DATA_4Q['cz_gates'],
         'o-', color='steelblue', linewidth=2, markersize=8, label='4-qubit (slope ≈3)')
ax2.plot(DATA_5Q['depths'], DATA_5Q['cz_gates'],
         's--', color='darkorange', linewidth=2, markersize=8, label='5-qubit (slope ≈4)')
x_fit = np.linspace(0, 22, 100)
for data in [DATA_4Q, DATA_5Q]:
    fit = np.polyfit(data['depths'], data['cz_gates'], 1)
    ax2.plot(x_fit, np.polyval(fit, x_fit), 'k:', alpha=0.4, linewidth=1.5)
ax2.set(xlabel='Circuit Depth (logical layers)', ylabel='CZ Gate Count',
        title='Gate Count Scaling with Depth')
ax2.set_xticks(DATA_4Q['depths'])
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
out = FIGURES_DIR / 'gate_count_analysis.png'
plt.savefig(out, dpi=300, bbox_inches='tight')
plt.savefig(RESULTS_DIR / 'gate_count_analysis.png', dpi=300, bbox_inches='tight')
print('Saved:', out)
plt.show()
```
